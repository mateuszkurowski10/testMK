```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Losowe dane
np.random.seed(42)
n = 100

x = np.arange(1, n + 1)

temperatura = 20 + x * 0.25 + np.random.normal(0, 3, n)
sprzedaz = 30 + x * 1.1 + np.random.normal(0, 12, n)
energia = 45 + x * 0.35 + np.random.normal(0, 5, n)

# Anomalie
sprzedaz[30] += 70
temperatura[65] += 15
energia[80] -= 20

# Wykres
plt.style.use("dark_background")

fig, ax = plt.subplots(figsize=(14, 8))
fig.patch.set_facecolor("#101020")
ax.set_facecolor("#181830")

ax.plot(x, temperatura, color="#00e5ff",
        linewidth=2.5, label="Temperatura")

ax.plot(x, sprzedaz, color="#ff00cc",
        linewidth=2.5, label="Sprzedaż")

ax.plot(x, energia, color="#aaff00",
        linewidth=2.5, label="Energia")

# Wypełnienie
ax.fill_between(x, temperatura, alpha=0.12,
                color="#00e5ff")

# Opisy
ax.set_title(
    "MEGA ANALIZA DANYCH – 100 POMIARÓW",
    fontsize=19, color="white", pad=20
)

ax.set_xlabel("Numer pomiaru", fontsize=13)
ax.set_ylabel("Wartość", fontsize=13)

ax.grid(alpha=0.2, linestyle="--")
ax.legend(fontsize=12, loc="upper left")

# Podsumowanie
print("ŚREDNIE WARTOŚCI")
print("Temperatura:", round(np.mean(temperatura), 2))
print("Sprzedaż:", round(np.mean(sprzedaz), 2))
print("Energia:", round(np.mean(energia), 2))

plt.tight_layout()
plt.savefig("mega_wykres.png", dpi=200)
plt.show()
```
