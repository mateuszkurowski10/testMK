```python
import tkinter as tk
import math

root = tk.Tk()
root.title("Machająca ręka 👋")
root.geometry("700x500")
root.resizable(False, False)

canvas = tk.Canvas(
    root, width=700, height=500,
    bg="#171735", highlightthickness=0
)
canvas.pack()

angle = 0
direction = 1

def transform(x, y, a):
    rad = math.radians(a)
    return (
        390 + x * math.cos(rad) - y * math.sin(rad),
        350 + x * math.sin(rad) + y * math.cos(rad)
    )

def draw_line(points, color, width):
    coords = []
    for x, y in points:
        X, Y = transform(x, y, angle)
        coords.extend([X, Y])

    canvas.create_line(
        *coords, fill=color, width=width,
        capstyle=tk.ROUND, joinstyle=tk.ROUND
    )

def animate():
    global angle, direction

    canvas.delete("all")

    # Tło
    canvas.create_rectangle(
        0, 0, 700, 500, fill="#171735", outline=""
    )

    # Gwiazdki
    for x, y in [(80, 80), (600, 100), (150, 300),
                 (550, 350), (620, 250), (300, 100)]:
        canvas.create_text(
            x, y, text="✨", font=("Arial", 18)
        )

    # Napis
    canvas.create_text(
        350, 75, text="CZEŚĆ! 👋",
        fill="#FFD166", font=("Arial", 36, "bold")
    )

    canvas.create_text(
        350, 125, text="Miłego dnia!",
        fill="white", font=("Arial", 22, "bold")
    )

    # Ramię
    draw_line([(0, 100), (0, 30)], "#FFB36B", 38)

    # Dłoń
    draw_line([(0, 30), (0, -35)], "#FFB36B", 43)

    # Palce
    draw_line([(-17, -48), (-25, -105)], "#FFB36B", 17)
    draw_line([(-7, -55), (-8, -125)], "#FFB36B", 17)
    draw_line([(5, -55), (10, -130)], "#FFB36B", 17)
    draw_line([(18, -48), (29, -110)], "#FFB36B", 17)

    # Kciuk
    draw_line([(-20, -20), (-48, -48)], "#FFB36B", 19)

    # Uśmiech na dole
    canvas.create_text(
        350, 450, text="Pozdrawiam wszystkich! ❤️",
        fill="#9BF6FF", font=("Arial", 19, "bold")
    )

    # Machanie ręką
    angle += direction * 4

    if angle >= 35:
        direction = -1
    elif angle <= -35:
        direction = 1

    root.after(30, animate)

animate()
root.mainloop()
```
