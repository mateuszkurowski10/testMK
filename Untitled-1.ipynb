```python
!pip install scipy
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generowanie danych o rozkładzie normalnym
np.random.seed(42)
dane = np.random.normal(loc=100, scale=15, size=1000)

# Obliczenie średniej i odchylenia standardowego
srednia = np.mean(dane)
odchylenie = np.std(dane)

# Histogram
plt.figure(figsize=(10, 6))
plt.hist(dane, bins=30, density=True, alpha=0.6,
         color='skyblue', edgecolor='black',
         label='Histogram danych')

# Krzywa rozkładu normalnego
x = np.linspace(min(dane), max(dane), 300)
y = norm.pdf(x, srednia, odchylenie)

plt.plot(x, y, color='red', linewidth=2.5,
         label='Rozkład normalny')

# Opisy wykresu
plt.title('Histogram i rozkład normalny')
plt.xlabel('Wartości')
plt.ylabel('Gęstość prawdopodobieństwa')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
